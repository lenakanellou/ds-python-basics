````python
import numpy as np

print("NumPy version:", np.__version__)
````

## 1. Python lists

A Python list is a general-purpose, ordered collection.

A list can contain values of different types:

````python
values = [1, 2, 3, 4, 5]

print(values)
print(type(values))
print(len(values))
````

## 2. NumPy arrays

A NumPy array is a collection designed for efficient numerical computation.

````python
array_values = np.array([1, 2, 3, 4, 5])

print(array_values)
print(type(array_values))
    print("Number of dimensions:", array_values.ndim)
print("Shape:", array_values.shape)
print("Number of elements:", array_values.size)
print("Data type:", array_values.dtype)
````

## 3. Multiplication: lists versus arrays

A Python list is repeated when multiplied by an integer. A NumPy array performs scalar multiplication element by element.

````python
print("List multiplied by 2:")
print(values * 2)

print()

print("Array multiplied by 2:")
print(array_values * 2)
````

## 4. Element-wise arithmetic

NumPy arithmetic operations are usually applied element by element.

````python
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
````

The same operations do not work element by element on ordinary lists.

For example, adding two lists concatenates them.

````python
list_a = [10, 20, 30, 40]
list_b = [1, 2, 3, 4]

print("List addition:", list_a + list_b)
print("Element-wise addition using a comprehension:", [x + y for x, y in zip(list_a, list_b)])
print("Element-wise addition using NumPy:", np.array(list_a) + np.array(list_b))
````

## 5. NumPy functions and summary statistics

NumPy provides mathematical and statistical functions that operate on complete arrays.

````python
scores = np.array([72, 81, 68, 90, 77, 85])

print("Number of scores:", scores.size)
print("Total:", scores.sum())
print("Mean:", scores.mean())
print("Median:", np.median(scores))
print("Minimum:", scores.min())
print("Maximum:", scores.max())
print("Standard deviation:", scores.std())
````

## 6. Boolean indexing

A comparison with an array produces an array of Boolean values. This Boolean array can be used as a mask.

````python
scores = np.array([45, 62, 78, 51, 89, 34, 73])

passed = scores >= 50

print("Mask:", passed)
print("Passing scores:", scores[passed])
print("Passing scores, shorter form:", scores[scores >= 50])
````

````python
scores_between_50_and_80 = scores[(scores >= 50) & (scores <= 80)]
    print(scores_between_50_and_80)
````

Use `&` for “and”, `|` for “or”, and `~` for “not”. Put each condition in parentheses.

## 7. Two-dimensional arrays

A two-dimensional array can represent a table or matrix. Here, rows represent students and columns represent assessments.

````python
marks = np.array([
    [72, 81, 69],
    [88, 91, 84],
    [65, 70, 74],
    [93, 89, 95]
])

print(marks)
print("Dimensions:", marks.ndim)
print("Shape:", marks.shape)
print("First student's first mark:", marks[0, 0])
print("Third student's second mark:", marks[2, 1])
print("Last student's marks:", marks[-1, :])
print("First assessment marks:", marks[:, 0])
````

## 8. Axis operations

For a two-dimensional array:

- `axis=0` operates down the rows and produces one result per column.
- `axis=1` operates across the columns and produces one result per row.

````python
print("Mean of each assessment:")
print(marks.mean(axis=0))

print()

print("Mean for each student:")
print(marks.mean(axis=1))

print()

student_totals = marks.sum(axis=1)
assessment_maxima = marks.max(axis=0)
student_means = marks.mean(axis=1)
best_student_index = student_means.argmax()

print("Student totals:", student_totals)
print("Assessment maxima:", assessment_maxima)
print("Best student row:", best_student_index)
````

## 9. Broadcasting

Broadcasting allows NumPy to combine arrays with compatible shapes.

Suppose we want to apply a different adjustment to each assessment:

- Assessment 1: add 2
- Assessment 2: add 0
- Assessment 3: subtract 1

````python
adjustments = np.array([2, 0, -1])
adjusted_marks = marks + adjustments

print(adjusted_marks)
````

The three-element adjustment array is applied to every row. This avoids writing a nested loop explicitly.

## 10. Data types

NumPy arrays normally contain values with a common data type.

````python
integers = np.array([1, 2, 3])
decimals = np.array([1.0, 2.0, 3.0])
mixed_numbers = np.array([1, 2.5, 3])

print(integers, integers.dtype)
print(decimals, decimals.dtype)
print(mixed_numbers, mixed_numbers.dtype)

mixed_list = [1, 2.5, "three"]
print(mixed_list)
print([type(value) for value in mixed_list])
````

## 11. Slices, views, and copies

A NumPy slice often produces a view into the original array rather than an independent copy.

````python
original = np.array([10, 20, 30, 40, 50])
section = original[1:4]

section[0] = 999

print("Original after changing the slice:", original)
print("Slice:", section)
````

````python
original = np.array([10, 20, 30, 40, 50])
section_copy = original[1:4].copy()

section_copy[0] = 999

print("Original:", original)
print("Independent copy:", section_copy)
````

Use `.copy()` when you want to modify a selected part without changing the original array.

## 12. Performance experiment

NumPy is often faster for large numerical computations because much of its work is implemented in compiled code.

This is a simple demonstration. Results depend on the operation, data size, hardware, and implementation.

````python
import time

n = 1_000_000
large_list = list(range(n))
large_array = np.arange(n)

start = time.perf_counter()
list_result = [value + 1 for value in large_list]
list_time = time.perf_counter() - start

start = time.perf_counter()
array_result = large_array + 1
array_time = time.perf_counter() - start

print(f"List comprehension: {list_time:.6f} seconds")
print(f"NumPy operation:    {array_time:.6f} seconds")
````

## 13. Mini-project: analyzing measurements

Rows represent observations and columns represent sensors.

````python
measurements = np.array([
    [20.1, 19.8, 20.4],
    [20.3, 20.0, 20.5],
    [19.9, 19.7, 20.1],
    [20.5, 20.2, 20.7],
    [20.0, 19.9, 20.3]
])

print(measurements)
print("Shape:", measurements.shape)
````

Complete the following tasks:

1. Calculate the mean for each sensor.
2. Calculate the standard deviation for each sensor.
3. Find the minimum and maximum value for each sensor.
4. Find all measurements above `20.4`.
5. Standardize each sensor using:

$$z = \\frac{x - \\mu}{\\sigma}$$

6. Verify that each standardized sensor has approximately mean zero.

````python
sensor_means = measurements.mean(axis=0)
sensor_stds = measurements.std(axis=0)
sensor_minima = measurements.min(axis=0)
sensor_maxima = measurements.max(axis=0)

print("Sensor means:", sensor_means)
print("Sensor standard deviations:", sensor_stds)
print("Sensor minima:", sensor_minima)
print("Sensor maxima:", sensor_maxima)
print("Values above 20.4:", measurements[measurements > 20.4])
````

````python
standardized = (measurements - sensor_means) / sensor_stds

print("Standardized measurements:")
print(standardized)
print()
print("Means after standardization:", standardized.mean(axis=0))
````

```python
class Queue:
    """A simple FIFO queue implemented with a Python list."""

    def __init__(self):
        """Create an empty queue."""
        self._items = []

    def enqueue(self, item):
        """Add an item to the back of the queue."""
        self._items.append(item)

    def dequeue(self):
        """Remove and return the item at the front of the queue."""
        if self.is_empty():
            raise IndexError("Cannot dequeue from an empty queue")

        return self._items.pop(0)

    def peek(self):
        """Return the front item without removing it."""
        if self.is_empty():
            raise IndexError("Cannot peek at an empty queue")

        return self._items[0]

    def is_empty(self):
        """Return True if the queue contains no items."""
        return len(self._items) == 0

    def __len__(self):
        """Return the number of items in the queue."""
        return len(self._items)

    def __repr__(self):
        """Return a useful representation of the queue."""
        return f"Queue({self._items!r})"
```

```python
tasks = Queue()

tasks.enqueue("Write report")
tasks.enqueue("Answer email")
tasks.enqueue("Attend meeting")

print(tasks)
print("Next task:", tasks.peek())
print("Number of tasks:", len(tasks))

print("Completed:", tasks.dequeue())
print("Completed:", tasks.dequeue())

print(tasks)
```

```python
queue = Queue()

assert queue.is_empty()
assert len(queue) == 0

queue.enqueue("A")
assert not queue.is_empty()
assert len(queue) == 1
assert queue.peek() == "A"
assert len(queue) == 1       # peek should not remove the item

queue.enqueue("B")
queue.enqueue("C")

assert len(queue) == 3
assert queue.peek() == "A"

assert queue.dequeue() == "A"
assert queue.dequeue() == "B"
assert queue.dequeue() == "C"

assert queue.is_empty()
assert len(queue) == 0

print("All queue tests passed.")
```

```python
empty_queue = Queue()

try:
    empty_queue.dequeue()
    assert False, "dequeue should raise IndexError on an empty queue"
except IndexError:
    pass

try:
    empty_queue.peek()
    assert False, "peek should raise IndexError on an empty queue"
except IndexError:
    pass

print("Error-handling tests passed.")
```

```python
def process_tasks(tasks):
    """Process every task and return the order in which tasks were processed."""
    completed = []

    while not tasks.is_empty():
        completed.append(tasks.dequeue())

    return completed


task_queue = Queue()

task_queue.enqueue("Check attendance")
task_queue.enqueue("Explain queues")
task_queue.enqueue("Give students an exercise")

completed_tasks = process_tasks(task_queue)

print(completed_tasks)
```

```python
task_queue = Queue()

task_queue.enqueue("Task 1")
task_queue.enqueue("Task 2")
task_queue.enqueue("Task 3")

result = process_tasks(task_queue)

assert result == ["Task 1", "Task 2", "Task 3"]
assert task_queue.is_empty()

print("Task-processing test passed.")
```
